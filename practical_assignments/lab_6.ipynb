### Протокол Диффи-Хеллмана


**Теоретическая часть:**
Протокол Диффи-Хеллмана позволяет двум сторонам получить общий секретный ключ через незащищённый канал связи. Безопасность протокола основана на сложности задачи дискретного логарифмирования.

**Основные формулы:**
- Открытые ключи: $A = g^a \mod p$, $B = g^b \mod p$
- Общий секрет: $K = B^a \mod p = A^b \mod p$

**Библиотеки:** `secrets`, `random`, `hashlib`

`AESGCM` — реализация аутентифицированного шифрования AES-GCM (обеспечивает и конфиденциальность, и целостность).

`serialization` — позволяет сохранять и передавать ключи в стандартизированных форматах (PEM, DER).

ec.SECP256R1() — стандартная NIST-кривая (secp256r1 / prime256v1), обеспечивающая ≈128-битную стойкость.
