### Изучение и реализация алгоритма RSA

**Теоретическая часть:**
RSA - асимметричный криптографический алгоритм, основанный на вычислительной сложности задачи факторизации больших чисел. Используется для шифрования и цифровых подписей.

**Основные формулы:**
- Модуль: $n = p \times q$
- Функция Эйлера: $\phi(n) = (p-1)(q-1)$
- Открытый ключ: $(e, n)$ где $1 < e < \phi(n)$ и $НОД(e, \phi(n)) = 1$
- Закрытый ключ: $(d, n)$ где $d = e^{-1} \mod \phi(n)$
- Шифрование: $c = m^e \mod n$
- Расшифрование: $m = c^d \mod n$

### Теоретическая справка к заданию 3.

#### Понятие «слабой реализации» RSA

«Слабая реализация» RSA — это реализация алгоритма, в которой нарушены криптографические практики (best practices), что делает систему уязвимой даже при математической корректности базового алгоритма. Основные причины слабой реализации:

- **Неправильный выбор параметров** — слишком малые `p`, `q` или `n`; предсказуемая или слабая генерация простых чисел.  
- **Отсутствие или некорректная реализация паддинга** (например, отсутствие OAEP/PKCS#1 v2), что позволяет применить атаки на «сырое» шифрование.  
- **Повторное использование параметров** (например, одинаковые `p` или `q` в разных ключах) или общий модуль `n` между пользователями.  
- **Использование слишком малого открытого (e) или секретного (d) экспонента**, что открывает конкретные методы атак (Hastad's Broadcast Attack, Wiener).  
- **Игнорирование защит от побочных каналов** (timing, power, fault), отсутствие проверок CRT — приводит к практическому компрометированию ключей.  
- **Несоблюдение стандартов и рекомендаций** (старые версии PKCS#1, нестандартизованные параметры).

#### Типы уязвимостей (и соответствующие атаки)
Безопасность RSA в основном связана с двумя задачами: факторизацией модуля $n = p \times q$ и сложностью инверсии $c \mapsto m^e \mod n$ (RSA-problem). При неправильной реализации появляются дополнительные пути атаки:

1. **Факторизация $n$ (малые $n$)**
   - Если $n$ слишком мало (например, учебные 256-512 бит) можно найти $p$ и $q$, восстановить $\phi(n) = (p-1)(q-1)$ и $d = e^{-1} \mod \phi(n)$.
   - Контрмера: использовать рекомендованные размеры ключей (2048 бит и выше).

2. **Общие делители между модулями (GCD-атака)**
   - Если два ключа используют один общий простой множитель (например, из-за плохого RNG при генерации простых чисел), то вычисление $\gcd(n_1, n_2)$ быстро выявит этот общий множитель.
   - Контрмера: надёжный CSPRNG при генерации простых чисел.

3. **Атаки при малом $e$ (Hastad's Broadcast Attack)**
   - Если одно и то же сообщение $m$ отправлено нескольким получателям с разными $n_i$, но с тем же малым $e$ (и без паддинга), то по $c_i = m^e \mod n_i$ можно восстановить $m$ с помощью CRT и восстановления корня.
   - Контрмера: использовать паддинг (OAEP), не использовать «сырые» $m^e$.

4. **Отсутствие паддинга / уязвимости PKCS#1 v1.5 (Bleichenbacher)**
   - Неправильный или устаревший паддинг открывает адаптивные атаки на расшифровку.
   - Контрмера: применять OAEP/PKCS#1 v2.2, защищённые парсеры, детектирование ошибок.

5. **Малый секретный экспонент $d$ (атака Wiener)**
   - Если $d$ достаточно мал (приблизительно $d < n^{0.25}$), то есть алгоритмы (Wiener), восстанавливающие $d$ по $(n, e)$.
   - Контрмера: гарантировать достаточную длину $d$, использовать безопасные генераторы ключей.

6. **Атаки по общему модулю (common modulus)**
   - Если один модуль $n$ используется для разных пар $e$, $d$ (редко, но встречается), это приводит к уязвимости.
   - Контрмера: каждому субъекту — свой $n$.

7. **Побочные каналы и ошибки CRT (fault attacks)**
   - Реализация с CRT без проверок может быть скомпрометирована при аппаратной ошибке (Bellcore attack и др.).
   - Контрмера: проверка результатов (verifying decrypt), защита от побочных каналов, проверка CRT.

#### Контрмеры (best practices)

- **Использовать проверенные библиотеки** и стандарты (OAEP, PKCS#1 v2.2).  
- **Минимальная длина ключа:** 2048 бит (минимальная рекомендуемая длина — 2048 бит. Для защиты особо чувствительной информации — 3072 или 4096 бит.).  
- **Надёжный генератор случайных чисел** (CSPRNG) при генерации `p,q` и при паддинге.  
- **Не использовать малые e или d**.  
- **Проверки при дешифровании:** в том числе проверка корректности CRT-результатов; защита от fault-атак.  
- **Защита от побочных каналов:** константное время, шум, аппаратные контрмеры.  
- **Избегать общего `n` и повторного использования простых** между ключами.
