### Реализация простой схемы аутентификации сообщения с использованием симметричного ключа


Аутентификация сообщения – это способ убедиться, что сообщение:
- было отправлено **тем, кем заявлено** (проверка подлинности);
- не было **изменено** при передаче (проверка целостности).

Наиболее распространённый способ реализации аутентификации в симметричных схемах — **HMAC (Hash-based Message Authentication Code)**. HMAC использует криптографическую хэш-функцию (например, SHA-256) и общий секретный ключ для вычисления и проверки кода.

### Пример реализация HMAC в Python:

```python
# Пример 1
import hmac, hashlib, secrets

key = secrets.token_bytes(32)
message = b'Message_to_send'

tag = hmac.new(key, message, hashlib.sha256).hexdigest()
print('HMAC:', tag)

check = hmac.compare_digest(tag, hmac.new(key, message, hashlib.sha256).hexdigest())
print('Проверка целостности:', check)
```
